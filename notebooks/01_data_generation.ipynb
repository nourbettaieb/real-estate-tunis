```python
import pandas as pd
import numpy as np
from datetime import datetime, timedelta
import random

print("=" * 70)
print("GENERATING REALISTIC TUNISIAN REAL ESTATE DATA")
print("=" * 70)

np.random.seed(42)
random.seed(42)

# ============================================
# DISTRICT DATA (Real Market Prices)
# ============================================

districts_data = {
    'Tunis Centre': {'avg_price': 250000, 'price_std': 100000, 'type': 'urban'},
    'La Marsa': {'avg_price': 280000, 'price_std': 120000, 'type': 'coastal'},
    'Ariana': {'avg_price': 180000, 'price_std': 80000, 'type': 'suburban'},
    'Ben Arous': {'avg_price': 160000, 'price_std': 70000, 'type': 'suburban'},
    'Rades': {'avg_price': 170000, 'price_std': 75000, 'type': 'suburban'},
    'Sfax Centre': {'avg_price': 140000, 'price_std': 60000, 'type': 'urban'},
    'Sousse Centre': {'avg_price': 160000, 'price_std': 70000, 'type': 'coastal'},
    'Hammamet': {'avg_price': 200000, 'price_std': 90000, 'type': 'coastal'},
    'Nabeul': {'avg_price': 130000, 'price_std': 55000, 'type': 'coastal'},
    'Gafsa': {'avg_price': 85000, 'price_std': 40000, 'type': 'rural'},
    'Kairouan': {'avg_price': 95000, 'price_std': 45000, 'type': 'rural'},
    'Kasserine': {'avg_price': 75000, 'price_std': 35000, 'type': 'rural'},
    'Bizerte': {'avg_price': 150000, 'price_std': 65000, 'type': 'coastal'},
    'Djerba': {'avg_price': 180000, 'price_std': 80000, 'type': 'coastal'},
    'El Kef': {'avg_price': 70000, 'price_std': 30000, 'type': 'rural'},
    'Jendouba': {'avg_price': 80000, 'price_std': 35000, 'type': 'rural'},
    'Mahdia': {'avg_price': 120000, 'price_std': 50000, 'type': 'coastal'},
    'Monastir': {'avg_price': 140000, 'price_std': 60000, 'type': 'coastal'},
}

# ============================================
# REALISTIC PROPERTY GENERATION
# ============================================

properties = []
n_properties = 200

for i in range(n_properties):
    
    # STEP 1: Select district
    district = np.random.choice(list(districts_data.keys()))
    district_info = districts_data[district]
    
    # STEP 2: Select property type
    property_type = np.random.choice(
        ['Apartment', 'Villa', 'Land', 'Duplex', 'House'],
        p=[0.45, 0.25, 0.15, 0.10, 0.05]  # Realistic distribution
    )
    
    # STEP 3: Generate REALISTIC area based on property type
    if property_type == 'Land':
        area = np.random.randint(300, 5000)
        bedrooms = 0
        price_multiplier = 1.0
    elif property_type == 'Apartment':
        area = np.random.randint(40, 180)
        bedrooms = np.random.choice([1, 2, 3, 4], p=[0.3, 0.4, 0.2, 0.1])
        price_multiplier = 1.0
    elif property_type == 'Villa':
        area = np.random.randint(150, 500)
        bedrooms = np.random.choice([3, 4, 5, 6], p=[0.3, 0.4, 0.2, 0.1])
        price_multiplier = 1.5  # Villas cost more per m²
    elif property_type == 'Duplex':
        area = np.random.randint(80, 250)
        bedrooms = np.random.choice([2, 3, 4], p=[0.3, 0.5, 0.2])
        price_multiplier = 1.2
    else:  # House
        area = np.random.randint(120, 350)
        bedrooms = np.random.choice([3, 4, 5], p=[0.4, 0.4, 0.2])
        price_multiplier = 1.3
    
    # STEP 4: Generate REALISTIC price
    # Base price per m²
    if district_info['type'] == 'urban':
        price_per_m2 = np.random.normal(2000, 500)
    elif district_info['type'] == 'coastal':
        price_per_m2 = np.random.normal(1800, 400)
    else:  # rural
        price_per_m2 = np.random.normal(800, 300)
    
    # Apply multiplier based on property type
    price_per_m2 = price_per_m2 * price_multiplier
    
    # Calculate total price
    base_price = int(area * price_per_m2)
    
    # Add some noise to make it realistic
    price = max(30000, int(base_price * np.random.uniform(0.85, 1.15)))
    
    # STEP 5: Generate realistic title
    if property_type == 'Land':
        title = f"Terrain {area}m² à {district}"
    else:
        title = f"{property_type} {bedrooms}P {area}m² - {district}"
    
    # STEP 6: Generate listing date (recent properties more common)
    if np.random.random() < 0.50:
        days_ago = np.random.randint(1, 7)
    elif np.random.random() < 0.60:
        days_ago = np.random.randint(7, 30)
    else:
        days_ago = np.random.randint(30, 91)
    
    date_listed = (datetime.now() - timedelta(days=days_ago)).strftime('%Y-%m-%d')
    
    # STEP 7: Generate additional realistic features
    has_parking = np.random.choice([True, False], p=[0.7, 0.3])
    has_garden = property_type in ['Villa', 'House'] and np.random.choice([True, False], p=[0.8, 0.2])
    condition = np.random.choice(['New', 'Excellent', 'Good', 'Fair'], p=[0.1, 0.3, 0.4, 0.2])
    
    # STEP 8: Add to list
    properties.append({
        'id': i + 1,
        'title': title,
        'price': price,
        'price_per_m2': round(price / area, 2),
        'location': f"{district} - Tunisia",
        'district': district,
        'area': area,
        'bedrooms': bedrooms,
        'property_type': property_type,
        'condition': condition,
        'has_parking': has_parking,
        'has_garden': has_garden,
        'date_listed': date_listed,
        'date_scraped': datetime.now().strftime('%Y-%m-%d')
    })

# ============================================
# CREATE DATAFRAME
# ============================================

df = pd.DataFrame(properties)

# ============================================
# SAVE TO CSV
# ============================================

df.to_csv('../data/raw_properties.csv', index=False)

print("\n" + "=" * 70)
print("✅ REALISTIC DATA GENERATED SUCCESSFULLY!")
print("=" * 70)

print(f"\n📊 STATISTICS:")
print(f"Total properties: {len(df)}")
print(f"Date range: {df['date_listed'].min()} to {df['date_listed'].max()}")

print(f"\n💰 PRICE ANALYSIS:")
print(f"  Average price: {df['price'].mean():,.0f} TND")
print(f"  Median price: {df['price'].median():,.0f} TND")
print(f"  Min price: {df['price'].min():,.0f} TND")
print(f"  Max price: {df['price'].max():,.0f} TND")

print(f"\n📐 AREA ANALYSIS:")
print(f"  Average area: {df['area'].mean():.1f} m²")
print(f"  Median area: {df['area'].median():.1f} m²")
print(f"  Min area: {df['area'].min():.1f} m²")
print(f"  Max area: {df['area'].max():.1f} m²")

print(f"\n🏠 PROPERTY TYPES:")
print(df['property_type'].value_counts())

print(f"\n📍 TOP 10 DISTRICTS BY LISTINGS:")
print(df['district'].value_counts().head(10))

print(f"\n💵 PRICE BY DISTRICT (Average):")
print(df.groupby('district')['price'].agg(['count', 'mean', 'min', 'max']).round(0))

print(f"\n🔑 CORRELATION: AREA vs BEDROOMS:")
print(f"  Apartments avg area: {df[df['property_type']=='Apartment']['area'].mean():.1f} m² ({df[df['property_type']=='Apartment']['bedrooms'].mean():.1f} bed avg)")
print(f"  Villas avg area: {df[df['property_type']=='Villa']['area'].mean():.1f} m² ({df[df['property_type']=='Villa']['bedrooms'].mean():.1f} bed avg)")
print(f"  Land avg area: {df[df['property_type']=='Land']['area'].mean():.1f} m² ({df[df['property_type']=='Land']['bedrooms'].mean():.1f} bed avg)")

print(f"\n✨ SAMPLE PROPERTIES (Shows realistic data):")
print(df[['title', 'price', 'area', 'bedrooms', 'property_type', 'district']].head(15))

print(f"\n✅ Data saved to: data/raw_properties.csv")
```
